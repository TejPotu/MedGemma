```python
import pandas as pd
import os
import shutil
import json

# Define paths
base_path = 'medical_datasets/whole_multicare_dataset'
output_dir = 'subset_dataset'
images_dir = os.path.join(output_dir, 'images')
data_dir = os.path.join(output_dir, 'data')

# Create directories
os.makedirs(images_dir, exist_ok=True)
os.makedirs(data_dir, exist_ok=True)
print(f"Created output directories: {images_dir}, {data_dir}")

# Load metadata
csv_path = os.path.join(base_path, 'captions_and_labels.csv')
cases_df = pd.read_csv(csv_path)

# Filter for chest x-rays
cxr_mask = (
    (cases_df['image_subtype'] == 'x_ray') &
    (cases_df['image_type'] == 'radiology') &
    (cases_df['caption'].str.contains(r'chest\s*x-?ray|cxr', case=False, na=False))
)
cxr_cases = cases_df[cxr_mask]
print(f"Found {len(cxr_cases)} matching cases.")

# Identify target filenames
target_filenames = set(cxr_cases['file'].values)
file_index = {}

# Indexing images
print("Indexing image files... (this may take a while)")
for root, dirs, files in os.walk(base_path):
    for file in files:
        if file in target_filenames:
            file_index[file] = os.path.join(root, file)
print(f"Indexed {len(file_index)} images.")

# Process and copy
count = 0
for index, row in cxr_cases.iterrows():
    filename = row['file']
    if filename in file_index:
        src = file_index[filename]
        dst = os.path.join(images_dir, filename)
        
        # Copy image
        shutil.copy2(src, dst)
        
        # Create JSON
        metadata = row.to_dict()
        metadata['local_image_path'] = os.path.join('images', filename)
        
        json_path = os.path.join(data_dir, f"{filename}.json")
        with open(json_path, 'w') as f:
            json.dump(metadata, f, indent=4)
        
        count += 1

print(f"Successfully processed {count} cases.")
```
